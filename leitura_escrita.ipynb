```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("LeituraEscrita").getOrCreate()
```

```python
df_videos = spark.read.option('header', 'true').option('inferSchema','true').csv('videos-stats.csv')
df_videos.show(8)
```

```python
df = spark.read.csv("videos-stats.csv", header=True, inferSchema=False)
df.show(8)
```

```python
df.printSchema()
```

```python
df = spark.read.csv("videos-stats.csv", header=True, inferSchema=True)
df.printSchema()
```

```python
df.write.mode("overwrite").option("header", "true").parquet("videos-parquet")
```

```python
df_videos_parquet = spark.read.parquet("videos-parquet")
df_videos_parquet.show()
```

```python
df_videos_parquet.write.mode("overwrite").saveAsTable("tabela_videos")
```

```python
spark.catalog.listTables()
```

```python
df_videos_tabela = spark.sql('select * from tabela_videos')

df_videos_tabela.show()
```

```python
df_comments = spark.read.csv("comments.csv", header=True, inferSchema=True)

df_comments.printSchema()
df_comments.show(8)
```

```python
df_comments.write.mode("overwrite").option("header", "true").parquet("comments-parquet")
```

```python
spark.stop()
```
